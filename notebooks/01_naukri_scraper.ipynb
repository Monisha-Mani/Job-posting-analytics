```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from bs4 import BeautifulSoup

import pandas as pd
import time
```

```python
roles = [

    "data-analyst",
    "business-analyst",
    "data-engineer",
    "data-scientist",
    "power-bi-developer",

    "python-developer",
    "java-developer",
    "full-stack-developer",

    "software-tester",
    "automation-tester",
    "selenium-tester",
    "qa-engineer",
    "sdet",

    "devops-engineer",
    "cloud-engineer",

    "cyber-security-analyst",

    "project-manager",

    "fresher-jobs",

    "associate-software-engineer"
]
```

```python
# Launch browser

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)
```

```python
# Main Scraping Loop

all_jobs = []

for role in roles:

    print(f"\nStarting role: {role}")

    for page in range(1, 21):

        if page == 1:

            url = f"https://www.naukri.com/{role}-jobs-in-india"

        else:

            url = f"https://www.naukri.com/{role}-jobs-in-india-{page}"


        print("Opening:", url)

        driver.get(url)

        time.sleep(4)

        soup = BeautifulSoup(
            driver.page_source,
            "html.parser"
        )

        job_cards = soup.find_all(
            "div",
            class_="srp-jobtuple-wrapper"
        )

        print("Jobs found:", len(job_cards))


        for card in job_cards:

            # Job Title
            title_tag = card.find("a", class_="title")

            # Company Name
            company_tag = card.find("a", class_="comp-name")

            # Experience
            exp_tag = card.find("span", class_="expwdth")

            # Salary
            sal_tag = card.find("span", class_="sal")

            # Location
            loc_tag = card.find("span", class_="locWdth")

            # Skills
            skill_tags = card.find_all(
                "li",
                class_="tag-li"
            )

            # Posted Date
            posted_tag = card.find(
                "span",
                class_="job-post-day"
            )

            # Description
            desc_tag = card.find(
                "span",
                class_="job-desc"
            )

            # Rating
            rating_tag = card.find(
                "span",
                class_="main-2"
            )

            # Reviews
            review_tag = card.find(
                "a",
                class_="review ver-line"
            )

            # Apply Link
            link = (
                title_tag["href"]
                if title_tag and title_tag.has_attr("href")
                else None
            )

            # Convert skills list to string
            skills = ", ".join(
                [
                    s.get_text(strip=True)
                    for s in skill_tags
                ]
            )

            # Store Data
            all_jobs.append({

                "Portal_Name": "Naukri",

                "Role_Search": role,

                "Job_Title":
                title_tag.get_text(strip=True)
                if title_tag else None,

                "Company_Name":
                company_tag.get_text(strip=True)
                if company_tag else None,

                "Experience_Required":
                exp_tag.get_text(strip=True)
                if exp_tag else None,

                "Salary":
                sal_tag.get_text(strip=True)
                if sal_tag else None,

                "Location":
                loc_tag.get_text(strip=True)
                if loc_tag else None,

                "Skills_Required":
                skills,

                "Posted_Date":
                posted_tag.get_text(strip=True)
                if posted_tag else None,

                "Job_Description":
                desc_tag.get_text(strip=True)
                if desc_tag else None,

                "Company_Rating":
                rating_tag.get_text(strip=True)
                if rating_tag else None,

                "Reviews_Count":
                review_tag.get_text(strip=True)
                if review_tag else None,

                "Job_Type": None,

                "Company_Size": None,

                "Industry": None,

                "Apply_Link": link
            })

print("Scraping completed")
```

```python
driver.quit()
```

```python
# creating data frame

df = pd.DataFrame(all_jobs)

print(df.shape)

df.head()
```

```python
#  Remove duplicate rows

df.drop_duplicates(inplace=True)


# Remove rows with missing critical fields

df.dropna(
    subset=[
        "Job_Title",
        "Company_Name",
        "Location"
    ],
    inplace=True
)


# Clean important text columns

text_cols = [

    "Portal_Name",

    "Role_Search",

    "Job_Title",

    "Company_Name",

    "Experience_Required",

    "Salary",

    "Location",

    "Skills_Required",

    "Posted_Date",

    "Job_Description",

    "Company_Rating",

    "Reviews_Count",

    "Job_Type",

    "Company_Size",

    "Industry",

    "Apply_Link"
]


for col in text_cols:

    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
    )


# Replace 'None' strings with actual null values

df.replace(
    ["None", "nan", ""],
    pd.NA,
    inplace=True
)


# Reset dataframe index

df.reset_index(
    drop=True,
    inplace=True
)


# Check final dataset shape

print(df.shape)


# Check missing values

print(df.isnull().sum())


# Preview dataset

df.head()
```

```python
# final cleaned dataset

df.to_csv(
    "../data/cleaned/naukri_final_dataset.csv",
    index=False
)

print("Final Naukri dataset exported successfully")
```

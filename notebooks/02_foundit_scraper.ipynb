```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from bs4 import BeautifulSoup

import pandas as pd
import time
```

```python
roles = [

    "data-analyst",
    "business-analyst",
    "data-engineer",
    "data-scientist",
    "power-bi-developer",

    "python-developer",
    "java-developer",
    "full-stack-developer",

    "software-tester",
    "automation-tester",
    "selenium-tester",
    "qa-engineer",
    "sdet",

    "devops-engineer",
    "cloud-engineer",

    "cyber-security-analyst",

    "project-manager",

    "fresher-jobs",

    "associate-software-engineer"
]
```

```python
# Launch browser

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)
```

```python
# Main Scraping Loop

all_jobs = []

for role in roles:

    print(f"\nStarting role: {role}")

    search_role = role.replace("-", "%20")


    for page in range(11, 21):

        url = f"https://www.foundit.in/srp/results?query={search_role}&page={page}&searchId="

        print("Opening:", url)

        driver.get(url)

        time.sleep(5)


        soup = BeautifulSoup(
            driver.page_source,
            "html.parser"
        )


        job_cards = soup.find_all(
            "div",
            class_="srpResultCardContainer"
        )


        print("Jobs found:", len(job_cards))


        for card in job_cards:


            # Job Title
            title_tag = card.find(
                "div",
                class_="jobTitle"
            )


            # Company Name
            company_tag = card.find(
                "div",
                class_="companyName"
            )


            # Posted Date
            posted_tag = card.find(
                "p",
                class_="timeText"
            )


            # Full Card Text
            full_text = card.get_text(
                " ",
                strip=True
            )


            # Experience
            exp = None

            if "Years" in full_text:

                exp = full_text


            # Location
            location = None

            if "India" in full_text:

                location = full_text


            # Apply Link
            link_tag = card.find("a")

            link = (
                link_tag.get("href")
                if link_tag else None
            )


            # Store Data
            all_jobs.append({

                "Portal_Name": "Foundit",

                "Role_Search": role,

                "Job_Title":
                title_tag.get_text(strip=True)
                if title_tag else None,

                "Company_Name":
                company_tag.get_text(strip=True)
                if company_tag else None,

                "Experience_Required":
                exp,

                "Salary": None,

                "Location":
                location,

                "Skills_Required": None,

                "Posted_Date":
                posted_tag.get_text(strip=True)
                if posted_tag else None,

                "Job_Description":
                full_text[:1000],

                "Company_Rating": None,

                "Reviews_Count": None,

                "Job_Type": None,

                "Company_Size": None,

                "Industry": None,

                "Apply_Link": link
            })

print("Scraping completed")
```

```python
driver.quit()
```

```python
# creating data frame

df = pd.DataFrame(all_jobs)

print(df.shape)

df.head()
```

```python
#  Remove duplicate rows

df.drop_duplicates(inplace=True)

print(df.shape)
```

```python
# Remove extra spaces

text_cols = [

    "Job_Title",
    "Company_Name",
    "Location",
    "Job_Description",
    "Posted_Date"
]

for col in text_cols:

    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
    )


# Replace invalid text

df.replace(
    ["None", "nan", ""],
    pd.NA,
    inplace=True
)


# Reset index

df.reset_index(
    drop=True,
    inplace=True
)

print(df.shape)
```

```python
# final cleaned dataset

df.to_csv(
    "../data/cleaned/foundit_final_dataset_2.csv",
    index=False
)

print("Final Foundit dataset exported successfully")
```

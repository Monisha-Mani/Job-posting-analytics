```python
# Step 1 - Import Required Libraries

import pandas as pd
import numpy as np
import re
from datetime import datetime
```

```python
# Step 2 - Load Master Combined Dataset

df = pd.read_csv("../data/cleaned/master_jobs_dataset.csv")

df.head()
```

```python
# Standardize Posted Date Column

from datetime import datetime, timedelta

today = datetime.today()

def convert_posted_date(value):

    if pd.isna(value):
        return pd.NaT

    value = str(value).lower().strip()

    try:

        if "day" in value:

            number = re.findall(r'\d+', value)

            if number:
                days = int(number[0])
                return today - timedelta(days=days)

        elif "week" in value:

            number = re.findall(r'\d+', value)

            if number:
                weeks = int(number[0])
                return today - timedelta(weeks=weeks)

        elif "month" in value:

            number = re.findall(r'\d+', value)

            if number:
                months = int(number[0])
                return today - timedelta(days=months * 30)

    except:
        return pd.NaT

    return pd.NaT


df["posted_date"] = df["posted_date"].apply(convert_posted_date)

print(df["posted_date"].head())
```

```python
# Step 3 - Standardize Job Roles


def standardize_role(title):

    title = str(title).lower()

    if "data analyst" in title:
        return "Data Analyst"

    elif "business analyst" in title:
        return "Business Analyst"

    elif "data scientist" in title:
        return "Data Scientist"

    elif "machine learning" in title or "ml engineer" in title:
        return "ML Engineer"

    elif "data engineer" in title:
        return "Data Engineer"

    elif "power bi" in title or "bi analyst" in title:
        return "BI Analyst"

    elif "python developer" in title:
        return "Python Developer"

    elif "selenium" in title or "sdet" in title or "automation tester" in title:
        return "QA Automation Engineer"

    elif "devops" in title:
        return "DevOps Engineer"

    elif "cloud engineer" in title:
        return "Cloud Engineer"

    elif "frontend" in title:
        return "Frontend Developer"

    elif "backend" in title:
        return "Backend Developer"

    elif "full stack" in title:
        return "Full Stack Developer"

    else:
        return "Others"


df["standardized_role"] = df["job_title"].apply(standardize_role)

print(df["standardized_role"].value_counts())
```

```python
# Step 4 - Skill Feature Engineering

skills = [
    "python",
    "sql",
    "power bi",
    "tableau",
    "excel",
    "selenium",
    "aws",
    "azure",
    "gcp",
    "java",
    "spark",
    "hadoop",
    "docker",
    "kubernetes",
    "jira",
    "jenkins",
    "rest api",
    "testng",
    "power apps"
]

for skill in skills:

    column_name = skill.replace(" ", "_") + "_flag"

    df[column_name] = df["skills_required"].str.lower().apply(
        lambda x: 1 if skill in str(x) else 0
    )

print(df.head())
```

```python
# Step 5 - Create Total Skills Count

skill_columns = [col for col in df.columns if "_flag" in col]

df["total_skills"] = df[skill_columns].sum(axis=1)

print(df["total_skills"].head())
```

```python
# Step 6 - Date Feature Engineering

df["posted_date"] = pd.to_datetime(
    df["posted_date"],
    errors="coerce",
    dayfirst=True
)

df["posting_month"] = df["posted_date"].dt.month_name()

df["posting_year"] = df["posted_date"].dt.year

df.head()
```

```python
# Step 7 - Job Description Features

df["jd_length"] = df["job_description"].astype(str).apply(len)

df["urgent_hiring_flag"] = df["job_description"].str.lower().apply(
    lambda x: 1 if "immediate joiner" in str(x) else 0
)

df.head()
```

```python
# Step 8 - Save Final Engineered Dataset

df.to_csv("../data/cleaned/final_engineered_jobs_dataset.csv", index=False)

print("Feature Engineering Completed")
```

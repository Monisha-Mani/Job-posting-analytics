```python
import pandas as pd

df = pd.read_csv("./data/raw/kaggle_jobs.csv")

print(df.head())

print(df.columns)

print(df.shape)
```

```python
df.columns = df.columns.str.lower().str.strip()

print(df.columns)
```

```python
df = df.rename(columns={
    'title': 'job_title',
    'jobid': 'job_id',
    'currency': 'currency',
    'jobuploaded': 'job_posted_date',
    'companyname': 'company_name',
    'tagsandskills': 'skills',
    'experience': 'experience',
    'salary': 'salary',
    'location': 'location',
    'companyid': 'company_id',
    'reviewscount': 'reviews_count',
    'aggregaterating': 'company_rating',
    'jobdescription': 'job_description',
    'minimumsalary': 'minimum_salary',
    'maximumsalary': 'maximum_salary',
    'minimumexperience': 'minimum_experience',
    'maximumexperience': 'maximum_experience'
})
```

```python
df.isnull().sum()
```

```python
df['company_name'] = df['company_name'].fillna('Unknown')

df['skills'] = df['skills'].fillna('Not Mentioned')

df['minimum_salary'] = df['minimum_salary'].fillna(0)

df['maximum_salary'] = df['maximum_salary'].fillna(0)

df['minimum_experience'] = df['minimum_experience'].fillna(0)

df['maximum_experience'] = df['maximum_experience'].fillna(0)

df['reviews_count'] = df['reviews_count'].fillna(0)

df['company_rating'] = df['company_rating'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df['experience'] = df['experience'].fillna(
    df['minimum_experience'].astype(str) + '-' +
    df['maximum_experience'].astype(str) + ' Years'
)
```

```python
df.isnull().sum()
```

```python
# removing duplicates

print("Before:", df.shape)

df = df.drop_duplicates()

print("After:", df.shape)
```

```python
# Standardising text column

text_columns = [
    'job_title',
    'company_name',
    'skills',
    'location'
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
# Converting them to lower case
for col in text_columns:
    df[col] = df[col].str.lower()
```

```python
# cleaning extra space
for col in text_columns:
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True)
```

```python
df.to_csv(
    "data/cleaned/kaggle_cleaned.csv",
    index=False
)

print("File Saved Successfully")
```

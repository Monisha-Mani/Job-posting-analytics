```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from bs4 import BeautifulSoup

import pandas as pd
import time
```

```python
roles = [

    "data-analyst",
    "business-analyst",
    "data-engineer",
    "data-scientist",
    "power-bi-developer",

    "python-developer",
    "java-developer",
    "full-stack-developer",

    "software-tester",
    "automation-tester",
    "selenium-tester",
    "qa-engineer",
    "sdet",

    "devops-engineer",
    "cloud-engineer",

    "cyber-security-analyst",

    "project-manager",

    "fresher-jobs",

    "associate-software-engineer"
]
```

```python
# Launch browser

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)
```

```python
# Main Scraping Loop

all_jobs = []

for role in roles:

    print(f"\nStarting role: {role}")

    search_role = role.replace("-", "%20")

    for page in range(1, 2):

        url = f"https://www.timesjobs.com/candidate/job-search.html?searchType=personalizedSearch&from=submit&txtKeywords={search_role}&txtLocation=&sequence={page}&startPage={page}"
        
        print(driver.current_url)

        driver.get(url)

        time.sleep(4)

        soup = BeautifulSoup(
            driver.page_source,
            "html.parser"
        )

        job_cards = soup.find_all(
            "div",
            class_="srp-card"
        )

        print("Jobs found:", len(job_cards))


        for card in job_cards:

            # Job Title
            title_tag = card.find("h2")


            # Company + Posted Date
            company_tag = card.find(
                "div",
                class_="text-xs text-gray-400 flex items-center"
            )

            company_text = (
                company_tag.get_text(strip=True)
                if company_tag else ""
            )


            # Split Company and Posted Date
            company_name = (
                company_text.split("|")[0]
                if "|" in company_text else company_text
            )

            posted_date = (
                company_text.split("|")[-1]
                if "|" in company_text else None
            )


            # Full Card Text
            full_text = card.get_text(" ", strip=True)


            # Experience
            exp = None

            for part in full_text.split():

                if "Yrs" in part:

                    exp = part
                    break


            # Salary
            salary = None

            if "LPA" in full_text:

                salary = (
                    full_text.split("LPA")[0][-20:]
                    + " LPA"
                )


            # Apply Link
            link_tag = card.find("a")

            link = (
                link_tag.get("href")
                if link_tag else None
            )


            # Store Data
            all_jobs.append({

                "Portal_Name": "TimesJobs",

                "Role_Search": role,

                "Job_Title":
                title_tag.get_text(strip=True)
                if title_tag else None,

                "Company_Name":
                company_name,

                "Experience_Required":
                exp,

                "Salary":
                salary,

                "Location": None,

                "Skills_Required": None,

                "Posted_Date":
                posted_date,

                "Job_Description":
                full_text[:1000],

                "Company_Rating": None,

                "Reviews_Count": None,

                "Job_Type": None,

                "Company_Size": None,

                "Industry": None,

                "Apply_Link": link
            })

print("Scraping completed")
```

```python
driver.quit()
```

```python
# creating data frame

df = pd.DataFrame(all_jobs)

print(df.shape)

df.head()
```

```python
#  Remove duplicate rows

df.drop_duplicates(inplace=True)


# Remove rows with missing critical fields

df.dropna(
    subset=[
        "Job_Title",
        "Company_Name"
        
    ],
    inplace=True
)


# Clean important text columns

text_cols = [

    "Portal_Name",

    "Role_Search",

    "Job_Title",

    "Company_Name",

    "Experience_Required",

    "Salary",

    "Location",

    "Skills_Required",

    "Posted_Date",

    "Job_Description",

    "Company_Rating",

    "Reviews_Count",

    "Job_Type",

    "Company_Size",

    "Industry",

    "Apply_Link"
]


for col in text_cols:

    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
    )


# Replace 'None' strings with actual null values

df.replace(
    ["None", "nan", ""],
    pd.NA,
    inplace=True
)


# Reset dataframe index

df.reset_index(
    drop=True,
    inplace=True
)


# Check final dataset shape

print(df.shape)


# Check missing values

print(df.isnull().sum())


# Preview dataset

df.head()
```

```python
# final cleaned dataset

df.to_csv(
    "../data/cleaned/timesjobs_final_dataset.csv",
    index=False
)

print("Final TimesJobs dataset exported successfully")
```

```python
from serpapi import GoogleSearch
import pandas as pd
import time
```

```python
API_KEY="YOUR_REAL_SERPAPI_KEY"
```

```python
queries = [
    "Data Analyst jobs in India",
    "Junior Data Analyst jobs in India",
    "Business Analyst jobs in India",
    "Power BI Analyst jobs in India",
    "SQL Data Analyst jobs in India",
    "Python Data Analyst jobs in India",
    "Remote Data Analyst jobs in India",
    "Fresher Data Analyst jobs in India"
]
```

```python
all_jobs = []

for query in queries:
    print(f"Fetching jobs for: {query}")

    params = {
        "engine": "google_jobs",
        "q": query,
        "hl": "en",
        "api_key": API_KEY
    }

    search = GoogleSearch(params)
    results = search.get_dict()

    jobs = results.get("jobs_results", [])

    print(f"Collected {len(jobs)} jobs")

    for job in jobs:
        all_jobs.append({
            "search_query": query,
            "job_title": job.get("title"),
            "company_name": job.get("company_name"),
            "location": job.get("location"),
            "via": job.get("via"),
            "description": job.get("description"),
            "thumbnail": job.get("thumbnail"),
            "job_highlights": str(job.get("job_highlights")),
            "detected_extensions": str(job.get("detected_extensions")),
            "job_id": job.get("job_id")
        })

    time.sleep(3)
```

```python
df_jobs = pd.DataFrame(all_jobs)
```

```python
print("Before dedup:", df_jobs.shape)

# remove duplicate jobs
df_jobs = df_jobs.drop_duplicates(subset=["job_id"])

print("After dedup:", df_jobs.shape)
```

```python
df_jobs.to_csv("serpapi_jobs_clean.csv", index=False)

print("Dataset saved successfully")
```
